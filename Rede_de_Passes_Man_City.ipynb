## Rede de passe 2019/2020

```python
import pandas as pd

# Carrega o CSV de eventos completos
df = pd.read_csv("man_city_2019_2020_events.csv")

# Filtra apenas eventos do Manchester City
df = df[df["team"] == "Man City"].copy()

# Ordena cronologicamente
df = df.sort_values(
    ["game_id", "period", "minute", "second"]
    ).reset_index(drop=True)

# Cria colunas com o próximo evento no mesmo jogo
df["next_game"] = df["game_id"].shift(-1)
df["receiver_id"] = df["player_id"].shift(-1)
df["receiver"] = df["player"].shift(-1)

# Mantém somente quando o próximo evento pertence ao mesmo jogo
df.loc[df["game_id"] != df["next_game"], ["receiver_id", "receiver"]] = None

# Filtra passes completos
passes = df[
        (df["type"] == "Pass") &
            (df["outcome_type"] == "Successful")
            ].copy()

# Remove passes sem receptor identificado
passes = passes.dropna(
                subset=["player_id", "receiver_id"]
                )

# Remove casos em que o jogador passa para si mesmo
passes = passes[passes["player_id"] != passes["receiver_id"]]

# Seleciona as colunas importantes
pass_links = passes[
    [
            "game",
                    "game_id",
                            "player_id",
                                    "player",
                                            "receiver_id",
                                                    "receiver",
                                                            "x",
                                                                    "y",
                                                                            "end_x",
                                                                                    "end_y"
                                                                                        ]
                                                                                        ]

                                                                                        # Salva todos os passes com receptor
pass_links.to_csv(
                                                                                            "man_city_2019_2020_pass_links.csv",
                                                                                                index=False
                                                                                                )

                                                                                                # Cria a matriz de conexões (arestas da rede)
pass_network = (pass_links.groupby(["game","game_id","player","receiver"                                                                                                                                                    ]
                                                                                                                                                                            )
                                                                                                                                                                                .size()
                                                                                                                                                                                    .reset_index(name="num_passes")
                                                                                                                                                                                    )

                                                                                                                                                                                    # Salva a rede agregada
pass_network.to_csv("man_city_2019_2020_pass_network.csv", index=False)
print(pass_network.head())
print(f"Número total de conexões: {len(pass_network)}")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mplsoccer import Pitch

# ==========================
# 1. Carregar passes
# ==========================
passes = pd.read_csv("man_city_2019_2020_pass_links.csv")

# ==========================
# 2. Definir o XI representativo
# ==========================
players = [
    "Ederson",
    "Kyle Walker",
    "Aymeric Laporte",
    "Nicolás Otamendi",
    "Oleksandr Zinchenko",
    "Rodri",
    "Kevin De Bruyne",
    "David Silva",
    "Riyad Mahrez",
    "Raheem Sterling",
    "Sergio Agüero"
]

# Filtra somente passes entre jogadores do XI
passes = passes[
    passes["player"].isin(players) &
    passes["receiver"].isin(players)
].copy()

# ==========================
# 3. Posições médias
# ==========================
origin = passes.groupby("player")[["x", "y"]].mean()

receive = (
    passes.groupby("receiver")[["end_x", "end_y"]]
    .mean()
)
receive.columns = ["x", "y"]

positions = {}

for player in players:
    coords = []

    if player in origin.index:
        coords.append(origin.loc[player].values)

    if player in receive.index:
        coords.append(receive.loc[player].values)

    if coords:
        positions[player] = np.mean(coords, axis=0)

positions = pd.DataFrame(positions, index=["x", "y"]).T


# ==========================
# 4. Criar arestas
# ==========================
edges = (
    passes
    .groupby(["player", "receiver"])
    .size()
    .reset_index(name="passes")
)

# Remove conexões muito fracas
threshold = edges["passes"].quantile(0.25)

edges = edges[edges["passes"] >= threshold]

print(f"Limiar escolhido: {threshold:.1f} passes")
print(edges.sort_values("passes", ascending=False))


# ==========================
# 5. Desenhar campo
# ==========================
pitch = Pitch(
    pitch_type="opta",
    pitch_color="#1B4332",
    line_color="white",
    linewidth=2
)

fig, ax = pitch.draw(figsize=(16, 10))


# ==========================
# 6. Desenhar conexões
# ==========================
max_passes = edges["passes"].max()

for _, edge in edges.iterrows():

    p1 = positions.loc[edge["player"]]
    p2 = positions.loc[edge["receiver"]]

    width = 1 + 10 * edge["passes"] / max_passes

    ax.plot(
        [p1.x, p2.x],
        [p1.y, p2.y],
        linewidth=width,
        color="#A8DADC",
        alpha=0.65,
        zorder=1,
        solid_capstyle="round"
    )


# ==========================
# 7. Jogadores
# ==========================
for player, pos in positions.iterrows():

    ax.scatter(
        pos.x,
        pos.y,
        s=450,
        color="#00B4D8",
        edgecolor="#023E8A",
        linewidth=2,
        zorder=3
    )

    # Sobrenome
    name = player.split()[-1]

    ax.text(
        pos.x,
        pos.y,
        name,
        color="white",
        fontsize=8,
        ha="center",
        va="center",
        fontweight="bold",
        zorder=4
    )


# ==========================
# 8. Título
# ==========================
ax.set_title(
    "Manchester City 2020/21 - Pass Network\nRepresentative XI (Agüero as Striker)",
    fontsize=18,
    color="white",
    weight="bold",
    pad=20
)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

def create_pass_network(input_file, output_prefix):
    # Carrega os eventos
    df = pd.read_csv(input_file)

    # Filtra apenas eventos do Manchester City
    df = df[df["team"] == "Man City"].copy()

    # Ordena cronologicamente
    df = df.sort_values(
        ["game_id", "period", "minute", "second"]
    ).reset_index(drop=True)

    # Identifica o próximo evento no mesmo jogo
    df["next_game"] = df["game_id"].shift(-1)
    df["receiver_id"] = df["player_id"].shift(-1)
    df["receiver"] = df["player"].shift(-1)

    # Remove ligações entre jogos diferentes
    df.loc[
        df["game_id"] != df["next_game"],
        ["receiver_id", "receiver"]
    ] = None

    # Seleciona apenas passes completos
    passes = df[
        (df["type"] == "Pass") &
        (df["outcome_type"] == "Successful")
    ].copy()

    # Remove passes sem receptor ou para si próprio
    passes = passes.dropna(
        subset=["player_id", "receiver_id"]
    )

    passes = passes[
        passes["player_id"] != passes["receiver_id"]
    ]

    # Passes individuais com origem e destino
    pass_links = passes[
        [
            "game",
            "game_id",
            "player_id",
            "player",
            "receiver_id",
            "receiver",
            "x",
            "y",
            "end_x",
            "end_y"
        ]
    ]

    # Salva os links de passes
    pass_links.to_csv(
        f"{output_prefix}_pass_links.csv",
        index=False
    )

    # Cria rede de passes por partida
    pass_network = (
        pass_links
        .groupby(
            [
                "game",
                "game_id",
                "player",
                "receiver"
            ]
        )
        .size()
        .reset_index(name="num_passes")
    )

    # Salva rede
    pass_network.to_csv(
        f"{output_prefix}_pass_network.csv",
        index=False
    )

    print(f"\n{output_prefix}")
    print(f"Jogos: {pass_links['game_id'].nunique()}")
    print(f"Total de passes: {len(pass_links)}")
    print(f"Conexões: {len(pass_network)}")

    return pass_links, pass_network


# ==========================
# Fase 2: até outubro de 2020
# ==========================

phase2_links, phase2_network = create_pass_network(
    "man_city_2020_2021_phase2_events.csv",
    "man_city_2020_2021_phase2"
)

# ==========================
# Fase 3: novembro de 2020 em diante
# ==========================

phase3_links, phase3_network = create_pass_network(
    "man_city_2020_2021_phase3_events.csv",
    "man_city_2020_2021_phase3"
)
```
